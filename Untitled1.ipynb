```python
!pip install tensorflow

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from google.colab import files
uploaded = files.upload()

treino = pd.read_csv('treino.csv')

print("Intervalo de IDs do treino: ", treino['id'].min(), "-", treino['id'].max())

x = treino.drop(columns=['id', 'target'])
y = treino['target']

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

x_train, x_val, y_train, y_val = train_test_split(x_scaled, y, test_size=0.2, random_state=42)

model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(x_train.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(5, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train, epochs=10, validation_data=(x_val, y_val))

loss, accuracy = model.evaluate(x_val, y_val)
print(f'Accuracy: {accuracy:.4f}')

uploaded = files.upload()
teste = pd.read_csv('teste.csv')
print("Intervalo de IDs do teste antes do ajuste: ", teste['id'].min(), "-", teste['id'].max())
teste = teste.sort_values(by='id')

x_teste = teste.drop(columns=['id'])
x_teste_scaled = scaler.transform(x_teste)

y_pred_teste = model.predict(x_teste_scaled)
y_pred_teste_classes = np.argmax(y_pred_teste, axis=1)

resultado = pd.DataFrame({'id': teste['id'], 'target': y_pred_teste_classes})
if resultado.shape[0] != 4501:
  print("ALERTA: Número incorreto de linhas no arquivo final", resultado.shape[0])

  ids_corretos = list(range(4501))
  df_new = pd.DataFrame({'id': ids_corretos[:4500], 'target': resultado['target']})
  df_corrigido = df_new

  if df_corrigido.shape[0] < 4501:
    df_corrigido.loc[4500] = [4500, df_corrigido['target'].mode()[0]]

    import os
    os.makedirs('/mnt/data', exist_ok=True)

  corrected_file_path = "/mnt/data/previsoes_corrigidas.csv"
  df_corrigido.to_csv(corrected_file_path, index=False)
  df_corrigido.head()
else:
  print("O arquivo contém 4.501 linhas, conforme esperado.")

resultado.to_csv('previsoes.csv', index=False)
files.download('previsoes.csv')
print("Arquivo previsoes.csv salvo com sucesso")
```
